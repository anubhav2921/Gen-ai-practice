```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("unclean_smartwatch_health_data.csv")

# Scatter plot
sns.scatterplot(
    data=df,
    x='Heart Rate (BPM)',
    y='Blood Oxygen Level (%)'
)

# Labels and title
plt.xlabel('Heart Rate (BPM)')
plt.ylabel('Blood Oxygen Level (%)')
plt.title('Heart Rate vs Blood Oxygen Level')

# Display graph
plt.show()
```
